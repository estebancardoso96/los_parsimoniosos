## Universo: estudiantes de tercer grado de educación media

### Bases de datos

#### Features: contexto

```python
df_contexto = pd.read_csv(r'https://www.ineed.edu.uy/wp-content/uploads/2023/08/Datos_Estudiante_CXTO.csv',  sep=";",
    encoding="cp1252")

df_contexto = df_contexto.query('IND_CXTO == 1')
```

#### Features: socioemocional

```python
df_socioemocional = pd.read_csv(r'https://www.ineed.edu.uy/wp-content/uploads/2023/08/Datos_Estudiante_Socioemocional.csv',  sep=";",
    encoding="cp1252")

df_socioemocional = df_socioemocional.query('IND_Socioemocional == 1')
```

#### Target: desempeño en matematica

```python
df_matematica = pd.read_csv(r'https://www.ineed.edu.uy/wp-content/uploads/2023/08/Datos_Estudiante_ITEMS_MAT.csv',  sep=";",
    encoding="cp1252")

df_matematica = df_matematica.query('IND_Estudiante_ITEMS == 1')
```

```python
id_contexto = df_contexto['AlumnoCodigoDes']
id_contexto = pd.DataFrame(id_contexto)

id_mat = df_matematica['AlumnoCodigoDes']
id_mat = pd.DataFrame(id_mat)

id_socioemocional = df_socioemocional['AlumnoCodigoDes']
id_socioemocional = pd.DataFrame(id_socioemocional)
```

```python
from functools import reduce

dfs = [id_contexto, id_mat, id_socioemocional]

df = reduce(
    lambda left, right: left.merge(right, on="AlumnoCodigoDes", how="inner"),
    dfs
)
```
